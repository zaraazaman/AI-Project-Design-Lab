# Object-Aware Illumination Direction Estimation

## 1. Project Overview

**Object-Aware Illumination Direction Estimation** is a computer vision project that aims to estimate the direction of the dominant light source in a scene from a single image.
The system will analyze visible **objects, shadows, and image geometry** to infer the direction from which the dominant illumination is coming.
The project uses the controlled illumination conditions of the MIT indoor scene dataset as ground truth for evaluating the estimated lighting direction.

## 2. Problem Statement

Determining the direction of illumination from a single photograph is difficult because the light source itself is often not visible.
The central research question of this project is:

> **Can computer vision estimate the direction of the dominant light source by analyzing objects, their visible shadows, and image geometry?**

The project approaches this problem by combining object information with shadow geometry to infer the direction of illumination.

### Why This Problem Matters

Understanding illumination direction from an image can provide useful information about the physical conditions under which a scene was captured. Estimating illumination from visual evidence also involves reasoning about the relationship between:

- **Objects and their geometry**
- **Visible shadows**
- **Shadow direction and displacement**
- **Image coordinates**
- **Lighting direction**

## 3. Core Objectives

The main objectives of the project are:

- **Detect objects** present in an input scene.
- Identify and analyze **visible shadows** associated with objects.
- Extract geometric information from objects and their shadows.
- Estimate the direction of the **dominant illumination source**.
- Represent the estimated illumination direction using the predefined lighting directions of the dataset.
- Compare the predicted lighting direction against the **known ground-truth illumination direction**.
- Evaluate the performance of the proposed computer vision approach.

### Project Pipeline

The intended high-level pipeline is:

```text
Input Image
     ↓
Object Detection
     ↓
Shadow Detection / Extraction
     ↓
Object-Shadow Geometry Analysis
     ↓
Illumination Direction Estimation
     ↓
Comparison with Ground Truth
     ↓
Evaluation
